```python
import os

# Shared Drives 확인
shared_path = '/content/drive/Shareddrives'
if os.path.exists(shared_path):
    print('[Shared Drives]')
    for name in os.listdir(shared_path):
        print(f'  {name}')
else:
    print('Shared Drives 없음')

# 내 드라이브 최상위 확인
print('\n[My Drive]')
for name in os.listdir('/content/drive/MyDrive'):
    print(f'  {name}')
```

```python
# 경로 확인 후 아래에 직접 입력
DATA_ROOT = ''  # 예: /content/drive/Shareddrives/pilliot/

assert DATA_ROOT, 'DATA_ROOT를 설정해주세요'
assert os.path.exists(DATA_ROOT), f'경로가 존재하지 않습니다: {DATA_ROOT}'

print('[DATA_ROOT 구조]')
for name in os.listdir(DATA_ROOT):
    print(f'  {name}')
```
